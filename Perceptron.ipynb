```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.animation import FuncAnimation
```

```python
#Define the Postivie and Negative classes
postitive_points= np.array([[-2, 3], [0, 1], [2,-1]])
negative_points= np.array([[-2, 1], [0, -1], [2, -3]])
```

```python
# Plot intial data points
plt.figure(figsize=(8,6))
plt.scatter(postitive_points[:, 0], postitive_points[:, 1], color='blue', marker='o', label='Positive Class')
plt.scatter(negative_points[:, 0], negative_points[:, 1], color='red', marker='x', label='Negative Class')
plt.legend(loc = 'best')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Initial Data Points')
plt.grid(True)
plt.show()
```

```python
# Combine all points and labels
x = np.vstack((postitive_points, negative_points))
y = np.array([1]*len(postitive_points) + [-1]*len(negative_points))
```

```python
x
y
```

```python
# percepton Algorithm
def perceptron(x,y,steps=100):
    theta = np.zeros(x.shape[1])
    theta0=0
    for t in range(steps):
        for i,xi in enumerate(x):
            if y[i]*(np.dot(theta,xi)+theta0)<=0:
                theta+=y[i]*xi
                theta0+=y[i]
                break

            # plot after each iteration
            plot_decision_boundary(theta, theta0,t+1)

    return theta, theta0        

    
```

```python
def plot_decision_boundary(theta, theta0, iteration):
    plt.figure(figsize=(8,6))
    plt.scatter(postitive_points[:, 0], postitive_points[:, 1], color='blue', marker='o', label='Positive Class')
    plt.scatter(negative_points[:, 0], negative_points[:, 1], color='red', marker='x', label='Negative Class')

    # Calculate decision boundary
    x_values = np.linspace(-3, 3, 2)
    if theta[1]!=0:
        y_values = -(theta[0]*x_values + theta0)/theta[1]
        plt.plot(x_values, y_values, color='green', label=f'Decision Boundary {iteration}')
    else:
        x_val = -theta0/theta[0]
        plt.axvline(x=x_val, color='green', label=f'Decision Boundary {iteration}')


    plt.xlim(-3, 3)
    plt.ylim(-4, 4) 
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.title(f'Decision Boundary after Step {iteration}')
    plt.legend(loc='best')
    plt.grid(True)
    plt.show()
```

```python
theta_final ,theta0_final= perceptron(x,y,10)
```
